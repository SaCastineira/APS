En este trabajo se programó la generacion de una señal sinusoidal mediante una funcion en python donde se definieron parametros para analizar su comportamiento.

### Definicion de la formula generadora de señal utilizada:

$$
x(t) = v_{\max}\sin(2\pi f t + \phi)+d_c 
$$

### Codigo utilizado 

```python
import matplotlib.pyplot as plt
import numpy as np

#%% Definiciones:
fs = 1000 #Hz
N = 1000 # cantidad de muestras   

#%% Funciones: "nn= cantidad de muestras, fs=frec de muestreo"
def mi_funcion_sen( vmax , dc , ff , ph, nn = N , fs = fs): 
   
   tt = np.arange(N)/fs # se mide en segundos porque es un vector de tiempo 

   xx = (vmax * np.sin( 2 * np.pi * ff * tt + ph)) + dc
  
   return(tt, xx)
```

Los parámetros utilizados son:

- $\mathrm{vmax}$: amplitud de la señal, en voltios (V).
- $\mathrm{dc}$: valor medio de la señal, en voltios (V).
- $\mathrm{ff}$: frecuencia de la señal, en hertz (Hz).
- $\mathrm{ph}$: fase de la señal, en radianes (rad).
- $\mathrm{nn}$: número de muestras, adimensional.
- $\mathrm{fs}$: frecuencia de muestreo, en hertz (Hz).
- $\mathrm{tt}$: tiempo correspondiente a las muestras, en segundos (s).
- $\mathrm{xx}$: valores de la señal, en voltios (V).

## Grafico base (Frecuencia =1,Vmax=25,dc=0)

```python
vmax =25 # Amplitud maxima 
dc = 0 # desplazamiento, + sube, - baja 
ff = 1# frecuencia variable 
ph = 0# fase ? radianes np.pi/2
tt, xx = mi_funcion_sen( vmax = vmax, dc = dc, ff = ff, ph=ph, nn = N, fs = fs)

plt.plot(tt, xx)
plt.grid(True)#genera una grilla
plt.title('Señal: senoidal' )
plt.xlabel('Tiempo [segundos]')
plt.ylabel('Amplitud [V]')
```

El gráfico muestra aproximadamente un ciclo completo de la señal senoidal. Como la frecuencia utilizada es de 1 Hz, el período de la señal es:

$$
T = \frac{1}{f} = 1 \; s
$$

El tiempo que transcurre entre dos muestras consecutivas se obtiene mediante:

$$
T_s = \frac{1}{f_s}
$$

En este caso se utiliza una frecuencia de muestreo de $f_s = 1000$ Hz y se toman $N = 1000$ muestras. Por lo tanto, la señal representada ocupa aproximadamente 1 segundo.

Como la primera muestra corresponde a $t = 0$, la última se encuentra en $t = 0.999$ s.

### Señal de 999 Hz

Para una señal de $f_f = 999\,\text{Hz}$ y una frecuencia de muestreo de $f_s = 1000\,\text{Hz}$, se supera la frecuencia de Nyquist:

$$
f_N = \frac{f_s}{2} = 500\,\text{Hz}
$$

Por lo tanto, aparece **aliasing**.

El período de muestreo es:

$$
T_s = \frac{1}{f_s} = 0.001\,\text{s}
$$

En ese tiempo, la señal de $999\,\text{Hz}$ recorre $0.999$ ciclos, es decir, casi un ciclo completo entre muestra y muestra.

Por eso, las muestras forman una señal aparente de:

$$
f_{alias} = f_s - f_f = 1000 - 999 = 1\,\text{Hz}
$$

La señal aparece invertida porque cada nueva muestra queda ligeramente atrasada respecto de la anterior, dando la impresión de que la senoide avanza en sentido contrario.

### Señales de 1001 Hz y 2001 Hz

Para $f_f = 1001\,\text{Hz}$, durante cada período de muestreo $T_s = 0.001\,\text{s}$, la señal recorre:

$$
1001 \cdot 0.001 = 1.001
$$

ciclos. Es decir, avanza un ciclo completo y una pequeña fracción adicional. Por aliasing, las muestras forman una señal aparente de:

$$
f_{alias} = f_f - f_s = 1001 - 1000 = 1\,\text{Hz}
$$

A diferencia del caso de $999\,\text{Hz}$, esta señal aparece en el sentido normal.

Para $f_f = 2001\,\text{Hz}$ ocurre lo mismo, ya que entre muestras se recorren $2.001$ ciclos. El muestreo vuelve a producir una señal aparente de $1\,\text{Hz}$.

En general, este comportamiento aparece para:

$$
f_f = kf_s + 1
$$

donde $k$ es un número entero positivo.
